## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*


### Research Question

Can we use historical page-level search and content signals to identify which web pages should be prioritized for a content-refresh review?

### Decision

The decision is which existing web pages an SEO team should inspect and potentially update first.

### Unit of Analysis

One individual web page.

### Output

A ranked list of pages with an estimated content-refresh opportunity score and supporting signals.

### Action

An SEO specialist reviews the highest-ranked pages, checks whether their information is outdated, and decides whether to update the content.

### Cost of a Wrong Recommendation

A false positive may cause a team to spend time updating a page that does not need changes, wasting resources.

A false negative may cause the team to overlook a page that could benefit from an update.

### Why Data and Machine Learning May Help

Manual rules can prioritize pages using a small number of thresholds. Machine learning may help identify more complex relationships between content age, search performance, and historical outcomes.

However, any model would need appropriate evaluation and validation before its scores could be trusted for real operational decisions.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python

# ML-02: Supporting evidence for the selected research lane

import pandas as pd
from pathlib import Path

# Locate the starter dataset
candidates = [
    Path("data/raw/content_refresh_anonymized.csv"),
    Path("../data/raw/content_refresh_anonymized.csv"),
    Path("/content/flyrank-ml-internship/data/raw/content_refresh_anonymized.csv"),
]

data_path = next((p for p in candidates if p.exists()), None)

if data_path is None:
    raise FileNotFoundError(
        "Starter dataset not found. Run the repository setup cell "
        "or clone the repository before running this analysis."
    )

df = pd.read_csv(data_path)

print("ML-02: Content Refresh Opportunity Research")
print("-------------------------------------------")
print("Total pages:", len(df))
print("Available columns:", list(df.columns))

required = [
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "ctr",
    "avg_position"
]

missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}")

# Supporting statistics
median_age = df["content_age_days"].median()
median_update = df["days_since_last_update"].median()
median_impressions = df["impressions_90d"].median()

print("\nThree supporting numbers:")
print(f"1. Median content age: {median_age:.1f} days")
print(f"2. Median days since last update: {median_update:.1f} days")
print(f"3. Median search impressions (90 days): {median_impressions:.1f}")

# Exploratory signals for potential refresh prioritization
old_pages = df[df["days_since_last_update"] > 180]

print("\nAdditional exploration:")
print(f"Pages not updated for over 180 days: {len(old_pages)}")
print(f"Percentage of pages not updated for over 180 days: {len(old_pages) / len(df) * 100:.1f}%")

print("\nInterpretation:")
print(
    "These descriptive statistics help assess whether the dataset "
    "contains variation relevant to content-refresh prioritization. "
    "They do not establish that updating content will improve rankings."
)
```
